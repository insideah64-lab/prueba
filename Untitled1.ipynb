```python
import json
import random
import os
from datetime import datetime, timedelta

class SkillNode:
    def __init__(self, name, description, level=1, requirements=None, system_type="Class"):
        self.name = name
        self.description = description
        self.level = level
        self.requirements = requirements or {}
        self.system_type = system_type
        self.sub_skills = {}

    def add_branch(self, skill_node):
        self.sub_skills[skill_node.name] = skill_node

    def to_dict(self):
        return {
            "name": self.name, "desc": self.description, "lvl": self.level,
            "sys": self.system_type, "branches": {k: v.to_dict() for k, v in self.sub_skills.items()}
        }

class AethelPlayer:
    def __init__(self, name, sex, origin, stats, initial_class, initial_job):
        self.name = name
        self.sex = sex
        self.origin = origin
        self.stats = stats
        self.player_class = initial_class
        self.player_job = initial_job
        self.class_lvl, self.class_xp = 1, 0
        self.job_lvl, self.job_xp = 1, 0
        self.hp, self.hunger, self.thirst = 100.0, 0.0, 0.0
        self.equipment = {"head": None, "chest": None, "hands": None, "legs": None, "weapon": None, "accessory": None}
        self.inventory = []
        self.is_alive = True

    def check_levelup_threshold(self, current_lvl):
        """Define la curva de experiencia necesaria para subir de nivel."""
        return current_lvl * 100

    def gain_xp(self, amount, system="Class"):
        msg = None
        if system == "Class":
            self.class_xp += amount
            threshold = self.check_levelup_threshold(self.class_lvl)
            if self.class_xp >= threshold:
                self.class_lvl += 1
                self.class_xp -= threshold
                msg = f"¡TRASCENDENCIA DE CLASE! {self.player_class} ha subido al Nivel {self.class_lvl}."
        else:
            self.job_xp += amount
            threshold = self.check_levelup_threshold(self.job_lvl)
            if self.job_xp >= threshold:
                self.job_lvl += 1
                self.job_xp -= threshold
                msg = f"¡MAESTRÍA PROFESIONAL! Tu oficio de {self.player_job} ha subido al Nivel {self.job_lvl}."
        return msg

    def award_xp_by_action(self, action, outcome):
        xp_gain = 25 if outcome == "ÉXITO" else 10
        combat_keywords = ["atacar", "matar", "defender", "luchar", "esquivar", "sangre", "golpear"]
        is_combat = any(kw in action.lower() for kw in combat_keywords)
        system = "Class" if is_combat else "Job"
        return self.gain_xp(xp_gain, system)
```

```python
class RPGItem:
    def __init__(self, name, item_type, weight, stats=None, rarity="Común", slot=None):
        self.name = name
        self.item_type = item_type  # 'arma', 'armadura', 'material', 'consumible'
        self.weight = weight
        self.stats = stats or {}
        self.rarity = rarity
        self.durability = 100.0
        self.slot = slot # 'weapon', 'chest', 'head', etc.

    def apply_rust(self, amount):
        """El Óxido degrada la durabilidad y efectividad del objeto."""
        self.durability = max(0, self.durability - amount)
        if self.durability < 20:
             for stat in self.stats: self.stats[stat] *= 0.8

class LootGenerator:
    @staticmethod
    def generate_random_item(tier=1):
        rarities = ["Común", "Raro", "Épico", "Único"]
        rarity_weights = [70, 20, 8, 2]
        chosen_rarity = random.choices(rarities, weights=rarity_weights)[0]

        # Definición de tipos de equipo y sus slots correspondientes
        equipment_data = {
            "arma": {"names": ["Daga de Hierro", "Espada de Ceniza", "Maza de Óxido"], "slot": "weapon"},
            "armadura_torso": {"names": ["Jubón de Cuero", "Coraza de Plomo", "Manto de Velo"], "slot": "chest"},
            "armadura_cabeza": {"names": ["Capucha de Tela", "Yelmo de Acero", "Corona de Espinas"], "slot": "head"}
        }

        i_type = random.choice(list(equipment_data.keys()))
        name = random.choice(equipment_data[i_type]["names"])
        slot = equipment_data[i_type]["slot"]

        base_stat = 5 * tier
        if chosen_rarity == "Raro": base_stat *= 1.5
        elif chosen_rarity == "Épico": base_stat *= 2
        elif chosen_rarity == "Único": base_stat *= 3

        stats = {"ataque": round(base_stat)} if "arma" in i_type else {"defensa": round(base_stat)}

        return RPGItem(name, i_type, weight=random.uniform(1.0, 8.0), stats=stats, rarity=chosen_rarity, slot=slot)
```

```python
def updated_calculate_derived_stats(self):
    # Penalización biológica
    penalty = 0.5 if (self.hunger > 85 or self.thirst > 85) else 1.0

    # Bonus de equipo
    armor_bonus = 0
    atk_bonus = 0
    for item in self.equipment.values():
        if item and hasattr(item, 'stats'):
            armor_bonus += item.stats.get('defensa', 0)
            atk_bonus += item.stats.get('ataque', 0)

    # Peso e inventario (RPGItem support)
    current_weight = sum(item.weight for item in self.inventory if hasattr(item, 'weight'))
    max_carry = 20 + (self.stats['Fuerza'] * 2)

    weight_penalty = 1.0
    if current_weight > max_carry:
        weight_penalty = 0.7

    return {
        "max_hp": round((50 + (self.stats['Resistencia'] * 5) + armor_bonus) * penalty, 2),
        "attack_power": round(((self.stats['Fuerza'] * 1.5) + atk_bonus) * weight_penalty, 2),
        "evasion": round((self.stats['Agilidad'] * 1.1) * weight_penalty * penalty, 2),
        "efficiency": round(penalty * weight_penalty, 2)
    }

AethelPlayer.calculate_derived_stats = updated_calculate_derived_stats
```

```python
class FactionSystem:
    def __init__(self):
        self.factions = {
            "Barones del Plomo": {"rep": 0, "stance": "Neutral", "desc": "Controlan el comercio de metales.", "influence": 0.8},
            "Culto del Óxido": {"rep": 0, "stance": "Neutral", "desc": "Adoran la entropía y la decadencia.", "influence": 0.4}
        }

    def update_reputation(self, faction_name, amount):
        if faction_name in self.factions:
            self.factions[faction_name]["rep"] += amount
            rep = self.factions[faction_name]["rep"]
            if rep > 50: self.factions[faction_name]["stance"] = "Aliado"
            elif rep < -50: self.factions[faction_name]["stance"] = "Hostil"
            else: self.factions[faction_name]["stance"] = "Neutral"
            return f"Tu posición con {faction_name} es ahora {self.factions[faction_name]['stance']}."
```

```python
def apply_environmental_wear(player, corruption_level):
    """Simula el efecto del Óxido del mundo sobre el equipo del jugador."""
    wear_msg = []
    for slot, item in player.equipment.items():
        if item and isinstance(item, RPGItem):
            wear_amount = corruption_level * 5.0
            item.apply_rust(wear_amount)
            if item.durability < 10:
                wear_msg.append(f"¡Tu {item.name} está a punto de romperse por el Óxido!")
    return "\n".join(wear_msg)
```

```python
class LegacyBridge:
    @staticmethod
    def process_succession(dead_player, chronicle):
        """Extrae el legado de un jugador muerto para el siguiente."""
        epitaph = f"Vivió como {dead_player.player_job} y cayó bajo el peso del Velo."
        chronicle.record_death(dead_player, epitaph)

        # Devolvemos un 'Starter Kit' basado en el progreso anterior
        legacy_bonus = {
            "Oro": dead_player.inventory.get("Oro", 0) * 0.2, # Solo se hereda el 20%
            "Esencia": dead_player.inventory.get("Esencia", 0)
        }
        return legacy_bonus
```

```python
class SkillGenerator:
    """Este módulo prepara la 'petición de semilla' para la IA."""

    @staticmethod
    def get_skill_generation_prompt(player, current_action):
        """
        Genera un prompt ultra-detallado para que la IA entienda
        CÓMO expandir el árbol fractal.
        """
        # Extraemos la 'Firma de Identidad' del jugador
        stats_context = {k: v for k, v in player.stats.items() if v > 5}

        prompt = f"""
        [SISTEMA DE GENERACIÓN FRACTAL DE HABILIDADES]

        CONTEXTO DEL JUGADOR:
        - Nombre: {player.name}
        - Senda Actual: {player.player_job}
        - Atributos Dominantes: {stats_context}
        - Acción Reciente: \"{current_action}\"

        INSTRUCCIONES PARA LA IA:
        1. RECURSIVIDAD: No crees una habilidad aislada. Crea una RAMA que sea hija de la habilidad '{player.player_job}'.
        2. TEMÁTICA FRACTAL: Si el jugador usa Inteligencia, la rama debe ser 'Arcana'. Si usa Fuerza, 'Física'.
        3. ESCALABILIDAD: Define un 'Tier' (1 a 5). Las de Tier 1 son básicas; las de Tier 5 son legendarias.
        4. EFECTOS MECÁNICOS: Asocia la habilidad a uno de los 12 atributos de Aethelraed.

        FORMATO DE RESPUESTA REQUERIDO (JSON):
        {{
            \"skill_name\": \"Nombre Único\",
            \"description\": \"Descripción inmersiva y gótica\",
            \"tier\": 1,
            \"stat_affinity\": \"Atributo Relacionado\",
            \"mana_cost\": 10,
            \"visual_theme\": \"Glitch/Neural/Void\"
        }}
        """
        return prompt
```

```python
class Monster:
    def __init__(self, name, hp, tier_ai, base_dmg, loot_table):
        self.name = name
        self.max_hp = hp
        self.hp = hp
        self.tier_ai = tier_ai
        self.base_dmg = base_dmg
        self.loot_table = loot_table
        self.is_variant = random.random() < 0.05

    def decide_tactic(self, player):
        """Lógica mecánica que la IA debe narrar."""
        if self.tier_ai == 'Instinto':
            return "Ataque frontal salvaje"
        elif self.tier_ai == 'Táctico':
            if player.stats['Agilidad'] < 10: return "Flanqueo y puñalada"
            return "Ataque calculado"
        elif self.tier_ai == 'Superior':
            return "Contragolpe y manipulación del entorno"
```

```python
class Market:
    def __init__(self, region_name):
        self.region_name = region_name
        self.inventory = {
            "Raciones": {"base": 10, "stock": 50},
            "Hierro": {"base": 40, "stock": 20},
            "Agua": {"base": 5, "stock": 100},
            "Madera": {"base": 15, "stock": 40}
        }
        self.world_modifier = 1.0

    def get_price(self, item_name, player_reputation):
        """Calcula el precio real usando la fórmula de escasez y reputación."""
        item = self.inventory.get(item_name)
        if not item: return 0

        scarcity = max(0.5, 2.0 - (item['stock'] / 100))
        rep_factor = max(0.7, 1.3 - (player_reputation / 100))

        final_price = item['base'] * scarcity * self.world_modifier * rep_factor
        return round(final_price, 2)
```

```python
class EventManager:
    def __init__(self):
        self.events = [
            {
                "name": "Niebla Errante",
                "description": "Una densa niebla desciende sobre la región.",
                "effect": {"world_modifier": 1.2},
                "duration": 2
            }
        ]
        self.active_events = []

    def trigger_random_event(self):
        if random.random() < 0.15 and not self.active_events:
            event = random.choice(self.events)
            self.active_events.append(event)
            return event
        return None

    def update_active_events(self):
        events_to_remove = []
        for event in self.active_events:
            event['duration'] -= 1
            if event['duration'] <= 0:
                events_to_remove.append(event)
        for event in events_to_remove:
            self.active_events.remove(event)
```
